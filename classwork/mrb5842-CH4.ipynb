```python
import numpy as np
```

```python
current_sum = 0

for n in range(5000):
    term = ((-1)**n) / (2*n + 1)
    current_sum += term
    
    calculated_pi = 4 * current_sum
    
    if (n + 1) % 100 == 0:
        relative_error = abs(calculated_pi - np.pi) / np.pi
        print(f"Term {n + 1}: {relative_error}")
```

```python
'''
The error decreases (gets smaller) as more terms are added, 
meaning the calculated value gets closer to the actual value of $\pi$. 
However, the convergence is very slow.
'''
```
