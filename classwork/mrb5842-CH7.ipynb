```python
import numpy as np

# Jacobi Method implementation
def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
            
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations

# Gauss-Seidel Method implementation
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i)) # Using already updated values
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n)) # Using old values
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
            
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations
```

```python
A1 = np.array([
    [0.582745, 0.48,   0.10,   0.00,     0.00    ],
    [0.48,     1.044129, 0.46,   0.10,     0.00    ],
    [0.10,     0.46,   1.10431,  0.44,     0.10    ],
    [0.00,     0.10,   0.44,   0.963889, 0.42    ],
    [0.00,     0.00,   0.10,   0.42,     0.522565]
], dtype=float)

b1 = np.array([1.162745, 2.084129, 2.20431, 1.923889, 1.042565], dtype=float)

# Initial guess of zeros [cite: 540]
x0_1 = np.zeros_like(b1)
tol = 1e-6
max_iters = 100

print("--- System 1 ---")
sol_j1, iters_j1 = jacobi(A1, b1, x0_1, tol, max_iters)
print(f"Jacobi Solution:\n {sol_j1}\n Iterations: {iters_j1}\n")

sol_gs1, iters_gs1 = gauss_seidel(A1, b1, x0_1, tol, max_iters)
print(f"Gauss-Seidel Solution:\n {sol_gs1}\n Iterations: {iters_gs1}")
```
