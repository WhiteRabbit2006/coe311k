```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
def monte_carlo_pi(num_samples):
    inside_circle = 0
    for _ in range(num_samples):
        x = random.random()
        y = random.random()
        if x**2 + y**2 <= 1.0:
            inside_circle += 1
    return (inside_circle / num_samples) * 4
```

```python
num_runs = 100
samples_per_run = 1000
```

```python
abs_errors = []
rel_errors = []

for _ in range(num_runs):
    pi_approx = monte_carlo_pi(samples_per_run)
    
    abs_err = abs(np.pi - pi_approx)
    abs_errors.append(abs_err)
    
    rel_err = abs((np.pi - pi_approx) / np.pi)
    rel_errors.append(rel_err)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

ax1.plot(range(num_runs), abs_errors, color='blue', marker='o', markersize=3, alpha=0.7)
ax1.set_title('Absolute Error per Run')
ax1.set_xlabel('Run Number')
ax1.set_ylabel('Absolute Error')
ax1.grid(True, linestyle='--', alpha=0.6)

ax2.plot(range(num_runs), rel_errors, color='orange', marker='s', markersize=3, alpha=0.7)
ax2.set_title('Relative Error per Run')
ax2.set_xlabel('Run Number')
ax2.set_ylabel('Relative Error')
ax2.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
