```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function
def f(x):
    return np.sin(x)

# Define the domain and number of points
a = 0             # Start of interval
b = 2 * np.pi     # End of interval
N = 50            # Number of points
x = np.linspace(a, b, N)

# Step size
h = (b - a) / (N - 1)
```

```python
# Central Difference Approximation for the derivative
central_diff = (f(x + h) - f(x - h)) / (2 * h)

# Forward Difference Approximation
forward_diff = (f(x + h) - f(x)) / h

# Backward Difference Approximation
backward_diff = (f(x) - f(x - h)) / h

# True derivative for comparison
true_derivative = np.cos(x)
```

```python
# Create a canvas with 4 subplots
plt.figure(figsize=(14, 10))

# 1. True Derivative
plt.subplot(2, 2, 1)
plt.plot(x, true_derivative, label='True: cos(x)', color='black', linewidth=2)
plt.title('True Derivative (Analytical)')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

# 2. Forward Difference
plt.subplot(2, 2, 2)
plt.plot(x, forward_diff, label='Forward Difference', color='red', linestyle='--')
plt.title('Forward Difference Approximation')
plt.xlabel('x')
plt.ylabel('Approximation')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

# 3. Backward Difference
plt.subplot(2, 2, 3)
plt.plot(x, backward_diff, label='Backward Difference', color='blue', linestyle='-.')
plt.title('Backward Difference Approximation')
plt.xlabel('x')
plt.ylabel('Approximation')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

# 4. Central Difference
plt.subplot(2, 2, 4)
plt.plot(x, central_diff, label='Central Difference', color='green', linestyle=':')
plt.title('Central Difference Approximation')
plt.xlabel('x')
plt.ylabel('Approximation')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

# Layout
plt.tight_layout()
plt.show()
```
