# Section 1: Introduction & Model Selection

## System Description
For this project, we are analyzing the dynamic response of a **Hobbico CS-6 UAV Servo Actuator**, which is an electro-mechanical component commonly used in Unmanned Aerial Vehicles (UAVs) to actuate flight control surfaces. The transient rotational response of the servo arm is heavily influenced by mechanical friction, inertia, and the internal motor's electrical characteristics. This behavior is effectively modeled as a **damped mechanical system**.

## Governing Equation
The unforced transient response of the servo actuator displacement can be modeled using the standard linear second-order ordinary differential equation (ODE) for a damped system:

$$ \frac{d^2y}{dt^2} + 2\zeta\omega\frac{dy}{dt} + \omega^2y = 0 $$

## Physical Meaning of Variables and Parameters
*   $y(t)$: The rotational displacement or angular response of the servo arm.
*   $t$: Time (seconds).
*   $\omega$: **Natural frequency** ($28.0 \text{ rad/s}$). It dictates the frequency at which the system would oscillate if there were no damping forces present.
*   $\zeta$: **Damping ratio** ($0.75$, dimensionless). It characterizes how rapidly the system's oscillations decay over time. Since $0 < \zeta < 1$, this specific servo actuator operates as an *underdamped* system, meaning it will exhibit some oscillation before settling at equilibrium.

## Conversion to a First-Order System
To solve this second-order ODE using standard numerical integration methods such as Euler's Forward and the Fourth-Order Runge-Kutta (RK4), we must convert it into a coupled system of two first-order ODEs. 

We introduce two state variables:
*   $y_1 = y(t)$ (Displacement)
*   $y_2 = \frac{dy}{dt}$ (Velocity)

Taking the time derivative of these state variables gives us our system:

1.  $\frac{dy_1}{dt} = y_2$
2.  $\frac{dy_2}{dt} = -2\zeta\omega y_2 - \omega^2 y_1$

This system can now be written in vector form $\frac{d\mathbf{Y}}{dt} = \mathbf{F}(t, \mathbf{Y})$ and integrated numerically.

## Initial Conditions
To simulate the transient behavior of the servo arm, we need appropriate initial conditions. We will assume the system is released from an initial offset angle while at rest.

*   $y(0) = y_1(0) = 1.0 \text{ rad}$: The servo arm starts at an initial displacement of 1 radian from equilibrium.
*   $y'(0) = y_2(0) = 0.0 \text{ rad/s}$: The servo arm is released from rest (zero initial velocity).

## Physical Scenario Description
The scenario being modeled is the free response of the Hobbico CS-6 servo actuator returning to its neutral equilibrium position after being subjected to an initial $1.0$ radian displacement. Because the damping ratio $\zeta = 0.75$ is less than $1$, the system is **underdamped**. We expect to see a relatively fast approach to the equilibrium state, accompanied by a slight overshoot and minimal oscillation before it fully settles.
