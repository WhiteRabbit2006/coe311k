```python
import numpy as np
import matplotlib.pyplot as plt

# The step sizes requested in the assignment
h_vals = [0.1, 0.01, 0.001, 0.00001, 0.0000001]
t_start = 0
t_end = 30
y_0 = 0

# The provided analytical solution
def exact_sol(t):
    return 3 - 0.998 * np.exp(-1000 * t) - 2.002 * np.exp(-t)

# The ODE function dy/dt
def f(t, y):
    return -1000 * y + 3000 - 2000 * np.exp(-t)

plt.figure(figsize=(12, 6))

for h in h_vals:
    # Calculate number of steps
    N = int(np.ceil((t_end - t_start) / h))
    
    # Note: For h = 0.0000001, N is 300 million. 
    # This might take 15-30 seconds to compute and uses a few GBs of RAM!
    t_arr = np.linspace(t_start, t_end, N + 1)
    y_arr = np.zeros(N + 1)
    y_arr[0] = y_0
    
    # Suppressing numpy overflow warnings because the larger h values 
    # (0.1 and 0.01) are mathematically unstable and will explode to infinity
    with np.errstate(over='ignore', invalid='ignore'):
        for i in range(N):
            y_arr[i+1] = y_arr[i] + h * f(t_arr[i], y_arr[i])
            
            # Break early if the solution explodes to infinity 
            # (Saves calculation time and prevents plotting NaN errors)
            if np.isinf(y_arr[i+1]) or np.isnan(y_arr[i+1]):
                y_arr[i+1:] = np.nan
                break
                
    # If the array is massive (like for h=1e-7), sub-sample it for plotting 
    # so matplotlib doesn't run out of memory and crash your browser.
    if N > 1000000:
        step = N // 10000
        plt.plot(t_arr[::step], y_arr[::step], label=f'h = {h}')
    else:
        plt.plot(t_arr, y_arr, label=f'h = {h}')

# Plot the exact solution
t_exact = np.linspace(t_start, t_end, 1000)
plt.plot(t_exact, exact_sol(t_exact), 'k--', linewidth=2, label='Analytical Solution')

# Formatting
# Stiff ODEs explode rapidly, so we must limit the y-axis to actually see the curve
plt.ylim(0, 4) 
plt.title("Euler's Forward vs Analytical Solution")
plt.xlabel('t')
plt.ylabel('y(t)')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()
```
